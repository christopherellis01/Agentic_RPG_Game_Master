```python
# ============================================================
# API smoke test for Agentic_RPG_Game_Master
# Run this from a notebook in the project environment.
# ============================================================

import subprocess
import sys
import importlib
from pathlib import Path

from fastapi.testclient import TestClient

api_path = Path(api.__file__)

compile_result = subprocess.run(
    [sys.executable, "-m", "py_compile", str(api_path)],
    capture_output=True,
    text=True,
)

if compile_result.returncode != 0:
    print("❌ src/api.py has a syntax error:")
    print(compile_result.stderr)
else:
    print("✅ src/api.py syntax check passed.")

# 2. Import/reload API module
try:
    import src.api as api
    importlib.reload(api)
    print("✅ src.api imported successfully.")
except Exception as e:
    print("❌ Could not import src.api:")
    raise e

# 3. Create FastAPI test client
client = TestClient(api.app)

# 4. Test /state
state_resp = client.get("/state")
print("\n/state status:", state_resp.status_code)
print(state_resp.json())

# 5. Test /reset
reset_resp = client.post("/reset")
print("\n/reset status:", reset_resp.status_code)
print(reset_resp.json())

# 6. Test combat turn through /turn
turn_resp = client.post(
    "/turn",
    json={
        "player_action": "I attack the goblin with my sword.",
        "target_npc_id": None,
    },
)

print("\n/turn status:", turn_resp.status_code)

if turn_resp.status_code != 200:
    print("❌ /turn failed:")
    print(turn_resp.text)
else:
    data = turn_resp.json()

    print("✅ /turn succeeded.")
    print("\nRoute:", data.get("route"))
    print("Route reason:", data.get("route_reason"))

    print("\nAgent activity:")
    for item in data.get("agent_activity", []):
        print("-", item.get("name"), "|", item.get("status"), "|", item.get("summary"))

    print("\nDialogue:")
    for line in data.get("dialogue", []):
        print(f"- {line.get('speaker')}: {line.get('said')}")

    print("\nRules payload:")
    print(data.get("rules"))

    # 7. Simple logic checks
    assert data.get("route") == "combat", "Expected route to be combat."
    assert data.get("rules") is not None, "Expected rules payload to exist."

    activity_names = [item.get("name") for item in data.get("agent_activity", [])]
    assert "rules" in activity_names, "Expected rules agent activity."
    assert "narrator" in activity_names, "Expected narrator activity."

    narrator_items = [
        item for item in data.get("agent_activity", [])
        if item.get("name") == "narrator"
    ]
    assert narrator_items[-1].get("status") == "done", "Expected narrator to be done."

    print("\n✅ Logic checks passed.")
```
